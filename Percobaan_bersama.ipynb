```python
def luas_lingkaran(r: int=2):
  """Menghitung luas lingkaran dengan rumus A = π * r^2"""
  area = 3.14 * (r ** 2)
  return area

print(f"area: {luas_lingkaran():.2f}")
print(f"area: {luas_lingkaran(8):.2f}")
print(f"area: {luas_lingkaran(12):.2f}")
```
